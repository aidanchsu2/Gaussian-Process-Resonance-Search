```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the combined table
df = pd.read_csv('combined_UL_brazil_1.96.csv')
mass = df['Mass']
obs  = df['Observed UL']
med  = df['Median UL']
lo1  = df['Lower 1σ']
hi1  = df['Upper 1σ']
lo2  = df['Lower 2σ']
hi2  = df['Upper 2σ']

plt.figure(figsize=(8,6))

# 2σ band (wider)
plt.fill_between(mass, lo2, hi2, alpha=0.5, label='±2σ band')
# 1σ band (narrower)
plt.fill_between(mass, lo1, hi1, alpha=0.5, label='±1σ band')

# Median expected and observed curves
plt.plot(mass, obs, linestyle='-',  label='Observed')

plt.plot(mass, med, linestyle='--', label='Median expected')

plt.yscale('log')
plt.xlabel('Mass [GeV]')
plt.ylabel('Upper limit')
plt.legend(loc='upper right')
plt.grid(which='both', linestyle=':')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Load & sort
obs   = pd.read_csv("combined_UL_fixed_ep.csv").sort_values("Mass")            # 0.0001 GeV steps
bands = pd.read_csv("combined_UL_brazil_fixed_ep.csv").sort_values("Mass")     # 0.001  GeV steps

x_obs = obs["Mass"].values
x_b   = bands["Mass"].values

# Interpolate a band column from 1 MeV grid -> 0.1 MeV grid
def interp_band(col):
    yb = bands[col].values
    # log-space interpolation is safer for ULs spanning decades
    yb = np.clip(yb, 1e-300, None)
    return np.exp(np.interp(x_obs, x_b, np.log(yb)))

lo2 = interp_band("Lower 2σ")
hi2 = interp_band("Upper 2σ")
lo1 = interp_band("Lower 1σ")
hi1 = interp_band("Upper 1σ")
med = interp_band("Median UL")
obs_ul = obs["Upper Limit"].values

# Restrict to overlap to avoid extrapolation
mask = (x_obs >= x_b.min()) & (x_obs <= x_b.max())

plt.figure(figsize=(8,6))
plt.fill_between(x_obs[mask], lo2[mask], hi2[mask], alpha=0.5, label="±2σ band")
plt.fill_between(x_obs[mask], lo1[mask], hi1[mask], alpha=0.5, label="±1σ band")
plt.plot(x_obs[mask], obs_ul[mask], "-",  label="Observed")
plt.plot(x_obs[mask], med[mask],   "--", label="Median expected")
plt.yscale("log")
plt.xlabel("Mass [GeV]")
plt.ylabel("Upper limit")
plt.legend(loc="upper right")
plt.grid(which="both", linestyle=":")
plt.tight_layout()
plt.show()
```
