```python
%load_ext autoreload
%autoreload 2
import sys
sys.path.append("..")
import gp
import pickle
import matplotlib.pyplot as plt
import numpy as np
from pandas import DataFrame
import openpyxl
import math
import scipy.integrate as integrate
# Making table for testing masses. Currently gives lots of inf values for unknown reason.
rebin = 1
#mass_hypotheses = np.arange(lower_bound,upper_bound,0.0001).tolist()

# Coefficients for the polynomial from the table for unsmeared (P0 to P4)
#coefficients = [0.00032, 0.019, -0.11, 1.39, -4.33]

#Smeared Coefficients
coefficients = [0.00038,0.041,-0.27,3.49,-11.11]

    
def sigma(mass, coeffs):
    #sigma is twice the mass resolution
    return sum(c * mass**i for i, c in enumerate(coeffs))

def sigmas(masses, coeffs):
    sigma_list = []
    for mass in masses:
        sigma_list.append(sum(c * mass**i for i, c in enumerate(coeffs)))
    return sigma_list
    
def make_table(masses, coeffs, lower_bound, upper_bound):
    sigma_list = sigmas(masses, coeffs)
    ul_list = []
    epsilon_squared_list = []
    for i in range(len(masses)):
        # Print blind range for debugging
        blind_range = (masses[i] - (1.5)*sigma_list[i], masses[i] + (1.5)*sigma_list[i])
        print(f"Mass: {masses[i]}, Sigma: {sigma_list[i]}, Blind Range: {blind_range}")
        if (blind_range[0] < lower_bound) or (blind_range[1] > upper_bound):
                print(f"upper limit not defined by mass resolution for  {masses[i]}")
                epsilon_squared = float("inf")
                upper_limit = float('inf')
                ul_list.append(upper_limit)
                epsilon_squared_list.append(epsilon_squared)
                continue
        try:
            # Create the model with the current mass and blind range
            m = gp.GaussianProcessModel(
                h = ('EventSelection_Data_10Percent.root', 'h_Minv_General_Final_1'),
                kernel = gp.kernels.WhiteKernel(noise_level=7e3) + gp.kernels.RBF(length_scale = 0.016)*gp.kernels.DotProduct(sigma_0 = 2.5e4),
                n_restarts_optimizer = 5,
                blind_range = blind_range,
                modify_histogram = gp._hist.manipulation.rebin_and_limit(rebin, lower_bound, upper_bound)
            )
            # Calculate upper limit
            upper_limit = m.upper_limit_in_blind_region()

            # Calculate epsilon squared
           #integral_range = (masses[i] - 2*sigma_list[i], masses[i] + 2*sigma_list[i])
            integral_range = (m.histogram.axes[0].index(masses[i] - 2*sigma_list[i]), m.histogram.axes[0].index(masses[i] + 2*sigma_list[i]) + 1)
            integral_bin = m.histogram.axes.widths[0][0]
            integral_size = (integral_range[1]-integral_range[0]) * integral_bin
            print(f"integral range:  {integral_range}")
            print(f"integral size:  {integral_size}")

            #integral = np.sum(m.histogram.values()[m.blind_range_indices[0]:m.blind_range_indices[1]+1])
            integral = np.sum(m.histogram.values()[integral_range[0]:integral_range[1]]) / integral_size
            print(f"integral for mass hypothesis {masses[i]}: {integral}")
            #print(f"blind range mass for hypothesis {masses[i]}: {m.histogram.axes.centers[0][m.blind_range_indices[0]]}, {m.histogram.axes.centers[0][m.blind_range_indices[1]]}")
            #print(f"integral range for mass {masses[i]}: {m.histogram.values()[m.blind_range_indices[0]:m.blind_range_indices[1]+1]}")
            frad = 18048.5*(masses[i]**5) - 10140.1*(masses[i]**4) + 2148.12*(masses[i]**3) - 212.211*(masses[i]**2) + 9.59977*masses[i] - 0.104070
            fsc = 1/137
            epsilon_squared = (2 * fsc * upper_limit)/(3 * math.pi * masses[i] * frad * integral)

            if ((masses[i] - (1.5)*sigma_list[i]) < lower_bound) or ((masses[i] + (1.5)*sigma_list[i]) > upper_bound):
                print(f"upper limit not defined by mass resolution for  {masses[i]}")
                epsilon_squared = float("inf")
                upper_limit = float('inf')
            
        except Exception as e:
            print(f"Error for mass {masses[i]}: {e}")
            upper_limit = float('inf')  # or a fallback value

        ul_list.append(upper_limit)
        epsilon_squared_list.append(epsilon_squared)

    return ul_list, sigma_list, epsilon_squared_list

#ul1, sl, es = make_table(mass_hypotheses, coefficients)

#df = DataFrame({'Mass': mass_hypotheses, 'Blind Range Size': [i*2 for i in sl], 'Upper Limit': ul1, 'Epsilon Squared': es})
#df.to_csv('ul_test.csv', index=False)
```

```python
import os
def truncate(f, n):
    '''Truncates/pads a float f to n decimal places without rounding'''
    s = '{}'.format(f)
    if 'e' in s or 'E' in s:
        return '{0:.{1}f}'.format(f, n)
    i, p, d = s.partition('.')
    return '.'.join([i, (d+'0'*n)[:n]])
    
# create output directory
output_dir = "ul_tables"
os.makedirs(output_dir, exist_ok=True)

# scan masses from 0.033 to 0.179 GeV in 1 MeV steps
for mass in np.arange(0.033, 0.180, 0.001):
    σ = sigma(mass, coefficients)
    lb = mass - 3*σ
    ub = mass + 3*σ
    lb_trunc = float(truncate(lb, 3))
    ub_trunc = float(truncate(ub, 3))
    to_search = np.arange(lb, ub, 0.0001)
    # make_table takes a list of masses + (coeffs, lower_bound, upper_bound)
    ul_list, sigma_list, eps2_list = make_table(to_search, coefficients, lb_trunc, ub_trunc)
    windows = [6 * x for x in sigmas(to_search, coefficients)]
    df = DataFrame({
        "Mass": to_search,
        "Window (GeV)": windows,
        "Upper Limit": ul_list,
        "Epsilon Squared": eps2_list
    })

    # write it out
    fname = f"UL_{mass:.3f}GeV.csv"
    path  = os.path.join(output_dir, fname)
    df.to_csv(path, index=False)

print(f"Wrote {len(os.listdir(output_dir))} files to ./{output_dir}/")
```
